```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
import numpy as np
import seaborn as sns
```

```python
events = pd.read_csv(
    "../data/test/test_FD001.txt", 
    header=None, 
    sep=" ",
)
events.columns = [
    "engine_id","cycle","setting1","setting2", "setting3",
    "s1","s2","s3","s4","s5","s6","s7","s8","s9","s10","s11","s12","s13","s14","s15","s16","s17","s18","s19","s20","s21",

]
cycles = pd.read_csv(
    "../data/RUL/RUL_FD001.txt", 
    header=None,
)
```

```python
true_rul = []

for engine_id in range(1, 101, 1):
    engine_events = events[events["engine_id"] == (engine_id)]
    print(engine_events)
    num_rows = engine_events.shape[0]
    ending_rul = cycles.iloc[engine_id-1,0]
    for row in range(0,num_rows):
        engine_cycle = engine_events.iloc[row, 1]
        true_rul.append(min((num_rows + ending_rul - engine_cycle), 130))
        
#print(true_rul)

events["true_RUL"] = true_rul
print(events)
```

```python
def pearson_viz(dataframe, col):
    corr_result=dataframe.corr(method = 'pearson')
    corr_result = corr_result.stack()
    corr_result = corr_result[((corr_result != 1.0)&(corr_result != -1.0))]
    corr_result = corr_result[((corr_result > 0.2)|(corr_result < -0.2))].round(2)
    df = pd.DataFrame(corr_result).reset_index()
    temp = df.loc[(df.level_0 == col)]
    cols = []
    for val in temp.level_1.values:
        cols.append(val)
    cols.append(col)
    sns.set(rc={'figure.figsize':(12,8),'ytick.labelsize':(12)})
    dataframe = dataframe[cols]
    correlation_matrix = dataframe.corr(method='pearson').round(2)
    sns.heatmap(data = correlation_matrix, annot = False, cmap = 'coolwarm_r')
    plt.title("Variable Correlation Heatmap\n", fontsize = 20)
    plt.show()
```

```python
pearson_viz(events, 'true_RUL')
```

```python
def comp_viz(dataframe, col):
    corr_result=dataframe.corr(method = 'pearson')
    corr_result = corr_result.stack()
    corr_result = corr_result[((corr_result != 1.0)&(corr_result != -1.0))]
    corr_result = corr_result[((corr_result > 0.2)|(corr_result < -0.2))].round(2)
    df = pd.DataFrame(corr_result).reset_index()
    temp = df.loc[(df.level_0 == col)]
    cols = []
    for val in temp.level_1.values:
        cols.append(val)
    cols.append(col)
    dataframe = dataframe[cols]
    sns.set(rc={'figure.figsize':(12,8),'ytick.labelsize':(12)})
    sns.pairplot(data = dataframe)
    plt.show()
```
